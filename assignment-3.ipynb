```python
import numpy as np

arr1 = np.random.randint(1, 101, size=(3, 3))
print( arr1)

min_val = arr1.min()
max_val = arr1.max()

normalized_arr = (arr1 - min_val) / (max_val - min_val)
print(normalized_arr)
```

```python
import numpy as np

arr2 = np.random.randint(1, 50, size=10)
print("Array:", arr2)

x = int(input("Enter a value X: "))

diff = np.abs(arr2 - x)
closest_index = diff.argmin()
closest_value = arr2[closest_index]

print("Closest value:", closest_value)
```

```python
import numpy as np

arr3 = np.random.randint(1, 100, size=(5, 5))
print("Array:\n", arr3)

sorted_arr = np.sort(arr3, axis=0)
second_largest = sorted_arr[-2]

print("Second-largest in each column:", second_largest)
```

```python
import numpy as np

arr4 = np.random.randint(1, 20, size=(6, 4))
print("Matrix:\n", arr4)

row_sums = arr4.sum(axis=1)

print("Indices of rows with an even sum:")
for i in range(len(row_sums)):
  if row_sums[i] % 2 == 0:
    print(i)
```

```python
import numpy as np

arr5 = np.random.randint(1, 51, size=30)
print("Original Array:\n", arr5)

for i in range(len(arr5)):
  if arr5[i] % 3 == 0 or arr5[i] % 5 == 0:
    arr5[i] = -1

print("Modified Array:\n", arr5)
```
